### import's

```python
import tensorflow as tf
from numpy import loadtxt
from keras.models import Sequential
from keras.layers import Dense, Input, Dropout
from tensorflow.keras.callbacks import EarlyStopping, LearningRateScheduler
from numpy import exp
from tensorflow.keras import Model

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.model_selection import GridSearchCV
from keras.wrappers.scikit_learn import KerasRegressor,KerasClassifier
```

### def model

```python
def exp_decay(self, epoch):
    initial_lrate = 0.001
    k = 0.1
    lrate = initial_lrate * exp(-k*epoch)
    return lrate

def model_mlp():
    input= Input(shape=(8,))
    x = Dense(10)(input)
    x = Dense(5)(x)
    x = Dropout(.2)(x)
    x = Dense(1)(x)
    model = Model(inputs=input, outputs=x)
    model.compile(loss='binary_crossentropy', metrics=['accuracy'])
    return model

def create_model(optimizer='Adam',epochs=100):

    model = Sequential()
    model.add(Dense(10,input_shape=(8,)))
    model.add(Dense(5))
    model.add(Dropout(.2))
    model.add(Dense(1))
    model.compile(loss='binary_crossentropy', metrics=['accuracy'], optimizer=optimizer)
    return model

def create_model_neurons(optimizer='Adam',layers=2, neurons=1,epochs=100):
    # create model
    model = Sequential()
    model.add(Dense(neurons, input_dim=8, kernel_initializer='uniform', activation='relu'))
    model.add(Dropout(0.2))
    for layer in range(layers):
        model.add(Dense(neurons, kernel_initializer='uniform', activation='relu'))
        model.add(Dropout(0.2))
    model.add(Dense(1, kernel_initializer='uniform', activation='sigmoid'))
    # Compile model
    model.compile(loss='binary_crossentropy', optimizer=optimizer, metrics=['accuracy'])
    return model
```

```python
# load the dataset
dataset = loadtxt('pima-indians-diabetes.csv', delimiter=',')
# split into input (X) and output (y) variables
X = dataset[:,0:8]
y = dataset[:,8]


X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
# create model

model = KerasClassifier(build_fn=create_model,verbose=1)

lr_schedule_1 = tf.keras.optimizers.schedules.ExponentialDecay(
    initial_learning_rate=0.1,
    decay_steps=100000,
    decay_rate=0.01,
    staircase=True)
lr_schedule_2 = tf.keras.optimizers.schedules.ExponentialDecay(
    initial_learning_rate=0.5,
    decay_steps=100000,
    decay_rate=0.1,
    staircase=True)

lr_schedule_3 = tf.keras.optimizers.schedules.ExponentialDecay(
    initial_learning_rate=0.5,
    decay_steps=100000,
    decay_rate=0.01,
    staircase=True)

#define the grid search parameters
optimizer = [tf.keras.optimizers.Adam(learning_rate=lr_schedule_1),
             tf.keras.optimizers.Adam(learning_rate=lr_schedule_2),
             tf.keras.optimizers.Adam(learning_rate=lr_schedule_3),]
epochs=[200,300]
param_grid = dict(optimizer=optimizer,epochs=epochs)

callabacks_list = [EarlyStopping(patience=50,monitor='loss')]

```

```python
grid = GridSearchCV(estimator=model, param_grid=param_grid, cv=3,verbose=1)
grid_result = grid.fit(X_train, y_train,callbacks=callabacks_list)
# summarize results
print("Best: %f using %s" % (grid_result.best_score_, grid_result.best_params_))
means = grid_result.cv_results_['mean_test_score']
stds = grid_result.cv_results_['std_test_score']
params = grid_result.cv_results_['params']
for mean, stdev, param in zip(means, stds, params):
    print("%f (%f) with: %r" % (mean, stdev, param))
```

```python
accuracy_score(y_test,grid_result.predict(X_test))
```
