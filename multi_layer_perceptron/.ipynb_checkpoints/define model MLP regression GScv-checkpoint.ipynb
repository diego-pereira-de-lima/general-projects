### import's

```python
import tensorflow as tf
import numpy as np
import pandas as pd
from numpy import loadtxt
from keras.models import Sequential
from keras.layers import Dense, Input, Dropout, BatchNormalization, Activation
from tensorflow.keras.callbacks import EarlyStopping, LearningRateScheduler
from numpy import exp
from tensorflow.keras import Model

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,mean_absolute_error
from sklearn.model_selection import GridSearchCV
from keras.wrappers.scikit_learn import KerasRegressor,KerasClassifier
```

### def model

```python
def exp_decay(self, epoch):
    initial_lrate = 0.001
    k = 0.1
    lrate = initial_lrate * exp(-k*epoch)
    return lrate


def create_model_neurons(optimizer='Adam',layers=0,dropout = 0.15, neurons=10,epochs=100):
    # create model
    model = Sequential()
    model.add(Dense(neurons, input_dim=13, kernel_initializer='uniform'))
    model.add(BatchNormalization())
    model.add(Activation('relu'))
    model.add(Dropout(dropout))
    for layer in range(layers):
        model.add(Dense(neurons))
        model.add(BatchNormalization())
        model.add(Activation('relu'))
        model.add(Dropout(dropout))
    model.add(Dense(1,activation='linear'))
    # Compile model
    model.compile(loss='mean_absolute_error',optimizer=optimizer, metrics=['mae'])
    return model
```

```python
from sklearn.datasets import load_boston
boston_dataset = load_boston()
X_train, X_test, y_train, y_test = train_test_split(boston_dataset.data, boston_dataset.target, test_size=0.33, random_state=42)
```

## Get best model

```python
# create model

model = KerasClassifier(build_fn=create_model_neurons, epochs=100, batch_size=10, verbose=1)

lr_schedule_1 = tf.keras.optimizers.schedules.ExponentialDecay(
    initial_learning_rate=0.1,
    decay_steps=100000,
    decay_rate=0.01)
lr_schedule_2 = tf.keras.optimizers.schedules.ExponentialDecay(
    initial_learning_rate=0.5,
    decay_steps=100000,
    decay_rate=0.1)
lr_schedule_3 = tf.keras.optimizers.schedules.ExponentialDecay(
    initial_learning_rate=0.05,
    decay_steps=100000,
    decay_rate=0.01)

#define the grid search parameters
optimizer = [tf.keras.optimizers.Adam(learning_rate=lr_schedule_1),
           #  tf.keras.optimizers.Adam(learning_rate=lr_schedule_2),
             tf.keras.optimizers.Adam(learning_rate=lr_schedule_3),]

epochs=[50,]
neurons = [ 5, 15]
layers =[0,2]
dropout=[.15, .25]
param_grid = dict(epochs=epochs,layers=layers,dropout=dropout,optimizer=optimizer,neurons=neurons)

callabacks_list = [EarlyStopping(patience=50,monitor='loss')]
# define the grid search parameters

grid = GridSearchCV(estimator=model, param_grid=param_grid, cv=3,verbose=1)
grid_result = grid.fit(X_train, y_train,callbacks=callabacks_list)
# summarize results
print("Best: %f using %s" % (grid_result.best_score_, grid_result.best_params_))
means = grid_result.cv_results_['mean_test_score']
stds = grid_result.cv_results_['std_test_score']
params = grid_result.cv_results_['params']
for mean, stdev, param in zip(means, stds, params):
    print("%f (%f) with: %r" % (mean, stdev, param))
```

```python
mlp_regressor=create_model_neurons()
mlp_regressor.fit(X_train, y_train,callbacks=callabacks_list)
mean_absolute_error(y_test,mlp_regressor.predict(X_test))
```

## RandomForestRegressor

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
param_grid = {'bootstrap': [True, False],
 'max_depth': [10, 20, 30],
 'max_features': ['auto', 'sqrt'],
 'min_samples_leaf': [1, 2, 4],
 'min_samples_split': [2, 5, 10],
 'n_estimators': [100,200, 300]}
```

```python
# Create a based model
rf = RandomForestRegressor()

grid_search = GridSearchCV(estimator = rf, param_grid = param_grid, 
                          cv = 3, n_jobs = -1, verbose = 2)
grid_search.fit(X_train, y_train)
mean_absolute_error(y_test,grid_search.predict(X_test))
```
