```python
def work(x):
    return x*x

with Pool(1) as p:
    results = p.map(work, range(5))
assert results == [x*x for x in range(5)]
```

```python
import multiprocessing

def write_pipe(pipe):
    print("write pipe started")
    p, _ = pipe
    for i in range(20):
        print("sending ", i)
        p.send(i)
    p.close()


def read_pipe(pipe):
    out, p = pipe
    out.close()
    try:
        while True:
            item = p.recv()
            print("recieved ", item)
    except EOFError:
        p.close()
```

```python
pipe = multiprocessing.Pipe(True)

p1 = multiprocessing.Process(target=write_pipe, args=(pipe,))
p2 = multiprocessing.Process(target=read_pipe, args=(pipe,))

p1.start()
p2.start()

pipe[0].close()

p1.join()
p2.join()

    
```

```python
from multiprocessing import Process, Pipe

def send_big_data(conn):
    data = b"a" * (1 << 30) # 1GB
    conn.send(data)
    conn.close() # eof

if __name__=="__main__":
    parent_conn, child_conn = Pipe()
    child = Process(target=send_big_data, args=(child_conn,))
    child.start()
    child_conn.close() # child must be the only with it opened
    print(len(parent_conn.recv()))
```
