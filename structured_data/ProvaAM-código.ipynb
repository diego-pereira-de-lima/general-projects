## Classificação

Considerando os dados presentes no arquivo class01.csv, treine o algoritmo Naive Bayes
Gaussiano utilizando a metodologia de validação cruzada holdout (utilize para treino as 350 primeiras
linhas e para validação as demais). Qual o valor da acurácia para a base de treino? Qual o valor da acurácia
para a base de validação? Faça o mesmo treinamento com a metodologia Leave-One-Out. Qual o valor da
acurácia média para a base de treino? Qual o valor da acurácia média para a base de validação?

```python
from sklearn.naive_bayes import GaussianNB
import pandas as pd
from sklearn.metrics import accuracy_score
data = pd.read_csv("class01.csv")
#hold-out
treino = data[:350]
y_treino = treino.target
validacao = data[350:]
y_validacao = validacao.target
treino = treino.drop('target',axis=1)
validacao = validacao.drop('target',axis=1)
gnb = GaussianNB()
gnb.fit(treino, y_treino)
gnb.score(treino, y_treino)
vali = gnb.predict(validacao)
accuracy_score(y_validacao, vali)

##leave-on-out
accuracia_treino = []
accuracia_validacao = []
from sklearn.model_selection import LeaveOneOut
y_data = data.target
data = data.drop('target', axis=1)
loo = LeaveOneOut()
for train_index, test_index in loo.split(data):
    X_train, X_validacao = data.loc[train_index], data.loc[test_index]
    y_train, y_validacao = y_data.loc[train_index], y_data.loc[test_index]
    gnb = GaussianNB()
    gnb.fit(X_train, y_train)
    accuracia_treino.append(gnb.score(X_train, y_train))
    vali = gnb.predict(X_validacao)
    accuracia_validacao.append(accuracy_score(vali, y_validacao))
np.mean(accuracia_treino)
np.mean(accuracia_validacao)
```

Considerando os dados presentes no arquivo class02.csv, treine o algoritmo 10-Nearest
Neighbors (KNN com e distância Euclidiana), utilizando a metodologia de validação cruzada k-fold com 10
folds não estratificados. Considere que a primeira pasta de validação seja formada pelas primeiras 10%
linhas do arquivo, que a segunda pasta de validação seja formada pelas 10% linhas seguintes, e assim por
diante, até atingir a última pasta, formada pelas 10% linhas finais da base. Qual o valor médio da acurácia
para a base de validação?

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import KFold
from sklearn.metrics import accuracy_score
data = pd.read_csv("class02.csv")
y_data = data.target
data = data.drop('target', axis=1)
accuracia_treino = []
accuracia_validacao = []
kf = KFold(n_splits=10, random_state=None, shuffle=False)
for train_index, test_index in kf.split(data):
    X_train, X_validacao = data.loc[train_index], data.loc[test_index]
    y_train, y_validacao = y_data.loc[train_index], y_data.loc[test_index]
    knn = KNeighborsClassifier(n_neighbors=10)
    knn.fit(X_train, y_train)
    accuracia_treino.append(knn.score(X_train, y_train))
    vali = knn.predict(X_validacao)
    accuracia_validacao.append(accuracy_score(vali, y_validacao))
np.mean(accuracia_treino) #0.8710370
np.mean(accuracia_validacao) #0.83
```

## Regressão

Considerando os dados presentes no arquivo reg01.csv, obtenha um modelo de regressão
linear com regularização L1 (LASSO com parâmetro de regularização igual a ) utilizando a metodologia
Leave-One-out. Qual o valor médio do Root Mean Squared Error (RMSE) para a base de treino? Qual o valor
médio do RMSE para a base de validação?

```python
from sklearn.model_selection import LeaveOneOut
from sklearn import linear_model
from sklearn.metrics import mean_squared_error
from math import sqrt
data = pd.read_csv("reg01.csv")
accuracia_treino = []
accuracia_validacao = []
y_data = data.target
data = data.drop('target', axis=1)
loo = LeaveOneOut()
for train_index, test_index in loo.split(data):
    X_train, X_validacao = data.loc[train_index], data.loc[test_index]
    y_train, y_validacao = y_data.loc[train_index], y_data.loc[test_index]
    reg = linear_model.Lasso(alpha = 1)
    reg.fit(X_train, y_train)
    tre = reg.predict(X_train)
    accuracia_treino.append(sqrt(mean_squared_error(y_train, tre)))
    vali = reg.predict(X_validacao)
    accuracia_validacao.append(sqrt(mean_squared_error(y_validacao, vali)))
np.mean(accuracia_treino)
np.mean(accuracia_validacao)
```
