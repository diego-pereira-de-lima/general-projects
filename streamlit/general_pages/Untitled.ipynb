```python
import altair as alt
import pandas as pd
```

```python
df=pd.read_excel('data/Base.xlsx')
```

```python
target = 'SAQ'
```

```python
df.info()
```

```python
df=df.groupby(by=['DATA']).mean()
```

```python
alt.Chart(df).mark_bar().encode(alt.X("MAE", bin=True),
    y='count()',)
```

```python
alt.Chart(df[df['AGENCIA']==1]).mark_line().encode(
    x='DATA:T', 
    y='MAE:N',
)

```
